```python
import pandas as pd
import nest_asyncio
import aiohttp
from understat import Understat
import asyncio

# Apply nest_asyncio to allow nested event loops for asynchronous functions
nest_asyncio.apply()

# Define an asynchronous function to fetch player data for a specific league and season from Understat
async def fetch_league_players(league, season):
    async with aiohttp.ClientSession() as session:
        understat = Understat(session)
        players_data = []  # List to store player data

        # Dictionary to hold UEFA rankings based on league and season
        uefa_rankings = {
            ("La Liga", 2017): 1, ("La Liga", 2018): 1, ("La Liga", 2019): 1, ("La Liga", 2020): 2,
            ("La Liga", 2021): 2, ("La Liga", 2022): 2,
            ("Bundesliga", 2017): 4, ("Bundesliga", 2018): 4, ("Bundesliga", 2019): 3, ("Bundesliga", 2020): 4,
            ("Bundesliga", 2021): 4, ("Bundesliga", 2022): 3,
            ("Ligue 1", 2017): 5, ("Ligue 1", 2018): 5, ("Ligue 1", 2019): 5, ("Ligue 1", 2020): 5,
            ("Ligue 1", 2021): 5, ("Ligue 1", 2022): 5,
            ("Serie A", 2017): 3, ("Serie A", 2018): 3, ("Serie A", 2019): 4, ("Serie A", 2020): 3,
            ("Serie A", 2021): 3, ("Serie A", 2022): 4,
            ("EPL", 2017): 2, ("EPL", 2018): 2, ("EPL", 2019): 2,
            ("EPL", 2020): 1, ("EPL", 2021): 1, ("EPL", 2022): 1
        }

        try:
            # Fetch players from the specified league and season
            league_players = await understat.get_league_players(league, season)

            # Convert the integer season to the "YY/YY" format (e.g., "17/18" for 2017)
            season_str = f"{str(season)[-2:]}/{str(season + 1)[-2:]}"

            for player in league_players:
                # Append relevant player data to the list
                players_data.append({
                    "ID": player["id"],
                    "Player Name": player["player_name"],
                    "Games": player["games"],
                    "Time": player["time"],
                    "Goals": player["goals"],
                    "xG": player["xG"],
                    "Assists": player["assists"],
                    "xA": player["xA"],
                    "Shots": player["shots"],
                    "Key Passes": player["key_passes"],
                    "Yellow Cards": player["yellow_cards"],
                    "Red Cards": player["red_cards"],
                    "Position": player["position"],
                    "Team Title": player["team_title"],
                    "npg": player["npg"],
                    "npxG": player["npxG"],
                    "xGChain": player["xGChain"],
                    "xGBuildup": player["xGBuildup"],
                    "League": league,
                    "UEFA league ranking": uefa_rankings.get((league, season), None),
                    "Season": season_str  # Insert season in "YY/YY" format
                })
            return players_data  # Return the list of player data
        except Exception as e:
            print(f"An error occurred for {league} in season {season}: {e}")
            return []  # Return an empty list in case of error

# Combine data for all leagues and seasons
async def main():
    leagues = ["Ligue 1", "Serie A", "Bundesliga", "La Liga", "EPL"]
    seasons = [2017, 2018, 2019, 2020, 2021, 2022]

    all_players_data = []  # List to accumulate data from all leagues and seasons

    for league in leagues:
        for season in seasons:
            # Fetch player data by awaiting the fetch_league_players function
            players_data = await fetch_league_players(league, season)

            # Append fetched data to the cumulative list if data exists
            if players_data:
                all_players_data.extend(players_data)
            else:
                print(f"No data available for {league} in season {season}.")

    # Convert the cumulative list to a DataFrame
    statistics_player = pd.DataFrame(all_players_data)

    # Replace 'EPL' with 'Premier League' in the League column
    statistics_player['League'] = statistics_player['League'].replace('EPL', 'Premier League')

    # Save the combined data to an Excel file
    statistics_player.to_excel('statistics_player.xlsx', index=False)

# Run the main function in an event loop
asyncio.run(main())
```

```python
import requests
import pandas as pd
from datetime import datetime

# Load Excel file
transfermarkt_table = pd.read_excel('transfermarkt_table.xlsx')

# Change Player ID type data column
transfermarkt_table['Player ID'] = transfermarkt_table['Player ID'].astype(pd.Int64Dtype())

# Lists to store results
Player_ID = []
Birth_Date = []
Height = []
Nationality = []
Foot = []

# Get unique player IDs
unique_player_id = transfermarkt_table['Player ID'].unique().tolist()

# Fetch player details from the Transfermarkt API
for player_id in unique_player_id:
    url = f'https://transfermarkt-api.fly.dev/players/{player_id}/profile'
    response = requests.get(url, headers={'accept': 'application/json'})

    if response.status_code == 200:
        data = response.json()

        Player_ID.append(data['id'])

        # Format the birth date
        date_str = data.get('dateOfBirth', None)
        if date_str is None:
            formatted_date = None
        else:
            try:
                # Convert birth date to dd.mm.yyyy format
                date_obj = datetime.strptime(date_str, "%b %d, %Y")
                formatted_date = date_obj.strftime("%d.%m.%Y")
            except ValueError:
                formatted_date = None

        Birth_Date.append(formatted_date)
        Height.append(data.get('height', None))

        # Check if 'citizenship' is a list and join it
        nationality_data = data.get('citizenship', None)
        Nationality.append(', '.join(nationality_data) if isinstance(nationality_data, list) else None)

        Foot.append(data.get('foot', None))
    else:
        # Append None if there's an error
        Player_ID.append(None)
        Birth_Date.append(None)
        Height.append(None)
        Nationality.append(None)
        Foot.append(None)

# Create a DataFrame with the results
details_table = pd.DataFrame({
    'Player ID': Player_ID,
    'Birth Date': Birth_Date,
    'Height': Height,
    'Nationality': Nationality,
    'Foot': Foot
})

# Split the Nationality column into two parts
nationalities_split = details_table['Nationality'].str.split(', ', expand=True)

# Add two columns for Nationalities even if some rows have only one nationality
details_table['Nationality1'] = nationalities_split[0]
details_table['Nationality2'] = nationalities_split[1] if nationalities_split.shape[1] > 1 else None

# Drop the original Nationality column and rename
details_table.drop(columns=['Nationality', 'Nationality2'], inplace=True)
details_table.rename(columns={'Nationality1': 'Nationality'}, inplace=True)

# Convert 'Player ID' column to Int64 type for details_table
details_table['Player ID'] = details_table['Player ID'].astype(pd.Int64Dtype())

# Merge details_table with transfermarkt_table
transfermarkt_table = pd.merge(transfermarkt_table, details_table, on='Player ID', how='left')

# Convert meters to centimeters in the Height column, if it is stored as a string in the format "x,y m"
def convert_height(value):
    try:
        value = str(value).replace('m', '').replace(',', '.')
        return float(value) * 100
    except ValueError:
        return None

transfermarkt_table['Height'] = transfermarkt_table['Height'].apply(convert_height)

# Save the merged data back to an Excel file
transfermarkt_table.to_excel('transfermarkt_table.xlsx', index=False)
```

```python
import requests
import pandas as pd
import time
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

# Load player data
players_data_set = pd.read_excel('players_data_set')

# Change Player ID type data column
players_data_set['Player ID'] = transfermarkt_table['Player ID'].astype(pd.Int64Dtype())

# List to store results
Player_ID = []
Value_Date = []
Value = []

# Get unique Player_ID
unique_player_id = players_data_set['Player ID'].unique().tolist()

# Configure request session with retries
session = requests.Session()
retry = Retry(
    total=5,
    backoff_factor=1,
    status_forcelist=[500, 502, 503, 504]
)
adapter = HTTPAdapter(max_retries=retry)
session.mount("https://", adapter)

# Fetch player value history from the Transfermarkt API
for player_id in unique_player_id:
    url = f'https://transfermarkt-api.fly.dev/players/{player_id}/market_value'

    try:
        response = session.get(url, headers={'accept': 'application/json'}, timeout=60)
        response.raise_for_status()
        data = response.json()
        print(f"Fetched data for Player ID {player_id}")

        if 'marketValueHistory' in data:
            for item in data['marketValueHistory']:
                if 'marketValue' in item and 'date' in item:
                    Player_ID.append(data.get('id', player_id))
                    Value_Date.append(item['date'])
                    Value.append(item['marketValue'])
        else:
            print(f"No market value history available for Player ID {player_id}")
    except requests.exceptions.ConnectTimeout:
        print(f"Timeout error for Player ID {player_id}")
    except requests.exceptions.HTTPError as err:
        print(f"HTTP error for Player ID {player_id}: {err}")
    except requests.exceptions.RequestException as err:
        print(f"Request error for Player ID {player_id}: {err}")

    time.sleep(1)  # Prevent API overload

# Create a DataFrame to store the details
player_values_table = pd.DataFrame({
    'Player ID': Player_ID,
    'Value Date': Value_Date,
    'Value': Value,
})

# Convert 'Value Date' to datetime
player_values_table['Value Date'] = pd.to_datetime(player_values_table['Value Date'], errors='coerce')
player_values_table.dropna(subset=['Value Date'], inplace=True)
player_values_table['Year'] = player_values_table['Value Date'].dt.year
player_values_table['Month'] = player_values_table['Value Date'].dt.month

# Filter data for the seasons 2017/18 to 2022/23 (years 2018-2023)
player_values_table = player_values_table[player_values_table['Year'].isin([2018, 2019, 2020, 2021, 2022, 2023])]

# Define preferred months for selecting market values (ordered by priority)
preferred_months = [6, 5, 7, 8, 4, 9]

# Function to get the value for preferred months or calculate yearly average
def get_value_for_preferred_months(group):
    preferred_values = group[group['Month'].isin(preferred_months)]

    if not preferred_values.empty:
        preferred_values = preferred_values.sort_values(
            by='Month',
            key=lambda x: x.map({6: 0, 5: 1, 7: 2, 8: 3, 4: 4, 9: 5})
        )
        best_value = preferred_values.iloc[0]
        return pd.Series({'Value': best_value['Value'], 'Value Date': best_value['Value Date'].strftime('%Y-%m-%d')})
    else:
        avg_value = group['Value'].mean()
        return pd.Series({'Value': avg_value, 'Value Date': str(group['Year'].iloc[0])})

# Apply function and reset index
final_values = player_values_table.groupby(['Player ID', 'Year']).apply(get_value_for_preferred_months).reset_index()

# Save results
final_values.to_excel('market_values.xlsx', index=False)
```

```python
import pandas as pd
import numpy as np

# Import data
players_data_set1 = pd.read_excel('players_data_set1.xlsx')
league_positions = pd.read_excel('League positions.xlsx')

# Keep only the necessary columns from league_positions and rename them
league_positions = league_positions[['Team', 'Season', 'Goals Scored', 'League Position']]
league_positions.rename(columns={'League Position': 'Team Position in League',
                                 'Goals Scored': 'Goals Scored by Team'}, inplace=True)

# Merge datasets based on team name and season
players_data_set1 = pd.merge(players_data_set1, league_positions,
                             left_on=['Team Title', 'Season'],
                             right_on=['Team', 'Season'],
                             how='left')

# Drop the redundant 'Team' column (since 'Team Title' is already in players_data_set1)
players_data_set1.drop(columns=['Team'], inplace=True)

# Avoid division by zero errors
players_data_set1['Goals Scored by Team'] = players_data_set1['Goals Scored by Team'].replace(0, np.nan)

# Calculate Goal Contribution Share (percentage of team goals contributed by a player)
players_data_set1['Goal Contribution Share'] = ((players_data_set1['Goals'] + players_data_set1['Assists']) /
                                                players_data_set1['Goals Scored by Team']) * 100

# Replace NaN values in Goal Contribution Share with 0 (if needed)
players_data_set1['Goal Contribution Share'].fillna(0, inplace=True)

# Drop the Goals Scored by Team column
players_data_set1.drop(columns=['Goals Scored by Team'], inplace=True)

# Save the merged dataset to a new CSV file
players_data_set1.to_excel("players_data_set1.xlsx", index=False)
```
